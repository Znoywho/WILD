```python
import numpy as np
import cv2 as cv
import matplotlib.pyplot as plt
AMBULADE_CONSTANT = np.array([0.299, 0.587, 0.114])
def convert2grey(img):
    return np.round(img[...,:3] @ AMBULADE_CONSTANT)
```

```python
img = cv.imread("grey_astronauts.jpg")
grey = convert2grey(img)
```

```python
theta = 45
transform_matrix = np.array(
    [[np.cos(theta), -np.sin(theta)],
    [np.sin(theta), np.cos(theta)]]
)

height, width = grey.shape[:2]

center = (width // 2, height // 2)

matrix = cv.getRotationMatrix2D(center, angle=45, scale=1.0)
rotated = cv.warpAffine(grey, matrix, (width, height))
ax = plt.subplots(1 ,2, figsize = (10, 10))[1] 
ax[0].imshow(grey, cmap ='grey')
ax[1].imshow(rotated, cmap = 'grey')
plt.show()
```

```python
sigmas = scale_pyramid()
pos = np.array(harris_laplace(grey, sigmas))
pos2 = np.array(harris_laplace(rotated, sigmas))
fig, ax = plt.subplots(1, 2, figsize=(10, 10))
ax[0].imshow(grey, cmap='grey')
ax[0].scatter(pos[:, 1].astype(np.int64), pos[:, 0].astype(np.int64), s=1, c="red")
ax[1].imshow(rotated, cmap='grey')
ax[1].scatter(pos2[:, 1].astype(np.int64), pos2[:, 0].astype(np.int64), s=1, c="red")
plt.show()
```

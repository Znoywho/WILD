```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np
img = cv.imread('images.jpg')
plt.imshow(img)
plt.show()
```

```python
LUMA = np.array([0.299, 0.587, 0.114]) # Luminanace
def convert2grey(img):
    return np.round(img[..., :3] @ LUMA).astype(np.uint8)
grey = convert2grey(img)
plt.imshow(grey, cmap= 'grey')
plt.show()
```

# Linear Model

```python
def mean_filter(img, k):
    img = img.astype(np.float64)
    H, W = img.shape
    filtered_img = np.zeros((H, W))
    for i in range(k, H - k):
        for j in range(k, W - k):
            window = img[i - k : i + k + 1, j - k : j + k + 1]
            filtered_img[i, j] = window.mean()
    return filtered_img.astype(np.uint8)

filtered_img = mean_filter(grey, 4)
plt.imshow(filtered_img, cmap='gray')
plt.show()
```

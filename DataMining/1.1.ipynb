# 6. Dọc và ghi file văn bản

|Chế độ|Ý nghĩa|
|---|---|
|`r`|Đọc file|
|`w`|Ghi file|
|`a`|Ghi nối tiếp|

```python
with open("danh_sach.txt", "w") as f:
    f.write("Le Thien Hao")
```

```python
with open("danh_sach.txt", "a") as f:
    f.write("\nNguyen Van A")
```

```python
with open("danh_sach.txt", 'r') as f:
    info = f.read()
    print(info)
```

```python
with open("danh_sach.txt", 'r') as f:
    for i in f:
        print(i.strip())
```

# 1. Nạp dự liệu iris vào bộ nhớ và hiển thị thông tin về số thuộc tính (dimensionality), số objects (size).

```python
from sklearn.datasets import load_wine
```

```python
wine = load_wine(as_frame=True)
```

```python
type(wine)
```

```python
print(wine)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
wine.data.hist()
```

```python
wine_data = wine.data
```

```python
wine_target = wine.target
```

```python
wine_target.value_counts()
```

# 3. Tách dữ liệu thành hai phần: training data(70%) và test data (30%

```python
from sklearn.model_selection import train_test_split
train_x, test_x, train_y, test_y = train_test_split(wine_data, wine_target, test_size=0.3)
```

```python
train_x
```

```python
len(train_y)
```

# 4. Chuẩn hóa dữ liệu với StandardScaler

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
fitted_train_x = scaler.fit_transform(train_x)
fitted_test_x = scaler.fit_transform(test_x)
```

# 5. Tạo và huấn luyện mô hình với training data

```python
from sklearn.tree import DecisionTreeClassifier
tree_clf = DecisionTreeClassifier(max_depth=2)
```

```python
tree_clf.fit(fitted_train_x, train_y)
```

```python
fitted_test_x = scaler.fit_transform(test_x)
```

```python
pred = tree_clf.predict(fitted_test_x)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, confusion_matrix, classification_report
```

```python
accuracy_score(test_y, pred)
```

```python
confusion_matrix(test_y, pred)
```

```python
print(classification_report(test_y, pred))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.set_printoptions(precision=2)

titles_options = [
    ("Confusion matrix, without normalization", None),
    ("Normalized confusion matrix", "true"),
]
for title, normalize in titles_options:
    disp = ConfusionMatrixDisplay.from_estimator(
        tree_clf,
        fitted_test_x,
        test_y,
        cmap=plt.cm.PuBu,
        normalize=normalize,
    )
    disp.ax_.set_title(title)

    print(title)
    print(disp.confusion_matrix)

plt.show()
```

```python
from sklearn.tree import export_graphviz
export_graphviz(
    tree_clf,
    out_file="wine.dot",
    feature_names=train_x.columns,
    rounded=True,
    filled=True
)
```

```python
lowercase_words = [w.lower() for w in words]
```

```python
unique_lowercase_words = set(lowercase_words)
```

```python
unique_lowercase_words
```

# 4. Đếm xem có bao nhiêu từ khác nhau (không phân biết hoa thường)

```python
len(unique_lowercase_words)
```

- có `7617` từ không phân biết hoa thường và duy nhất

```python
word_counts = {}
for word in lowercase_words:
    word_counts[word] = word_counts.get(word, 0) + 1

print(word_counts)
```

```python
sorted_word_counts = dict(
    sorted(
        word_counts.items(), 
        key=lambda item:item[1], 
        reverse=True
    )
)
```

```python
sorted_word_counts
```
